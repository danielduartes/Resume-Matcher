### Arquivo utilizado para converter o formato de arquivo do currículo, geralmente feito em PDF ###

```python
import fitz 
from pathlib import Path
```

```python
doc = fitz.open("entrada/curriculo/Currículo (2).pdf")
texto = "\n".join(pagina.get_text() for pagina in doc)
```

```python
Path("entrada/curriculo/curriculo.txt").write_text(texto, encoding="utf-8")
print(texto[:600])
```
